# GitPulse EDA and forecast check

Explore `daily_facts`, plot a topic, and inspect `ds/results.json` after training.

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sqlalchemy import text

from gitpulse.db import connect

with connect() as conn:
    facts = pd.read_sql(text("SELECT * FROM daily_facts ORDER BY day"), conn)

facts.tail()
```

```python
topic = "llm"
part = facts[facts["topic"] == topic].copy()
part["day"] = pd.to_datetime(part["day"])

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(part["day"], part["stars"], label="stars")
ax.plot(part["day"], part["pushes"], label="pushes", alpha=0.7)
ax.set_title(f"{topic} daily activity")
ax.legend()
plt.show()
```

```python
path = Path("ds/results.json")
if path.exists():
    print(json.dumps(json.loads(path.read_text()), indent=2))
else:
    print("Run: python -m ds.train")
```
